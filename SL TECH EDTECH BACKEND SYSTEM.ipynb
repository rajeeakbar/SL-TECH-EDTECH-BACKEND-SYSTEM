```python

# ==========================================================
# USER CLASS
# ==========================================================

class User:
    """
    Parent class for all users.

    Stores basic information common to learners
    and instructors.
    """

    def __init__(self, user_id, name, email, password):

        self.user_id = user_id
        self.name = name
        self.email = email
        self.password = password

    def update_email(self, new_email):
        """
        Updates user email address.
        """

        self.email = new_email

    def update_password(self, new_password):
        """
        Updates user password.
        """

        self.password = new_password

    def validate_credentials(self, email, password):
        """
        Validates login credentials.
        """

        return (
            self.email == email
            and self.password == password
        )

    def display_user(self):

        return (
            f"ID: {self.user_id}, "
            f"Name: {self.name}, "
            f"Email: {self.email}"
        )
```

```python

# ==========================================================
# LEARNER CLASS
# ==========================================================

class Learner(User):
    """
    Learner inherits from User.

    Additional functionality:
    - Enroll into course
    - Drop course
    """

    def __init__(self,
                 user_id,
                 name,
                 email,
                 password):

        super().__init__(
            user_id,
            name,
            email,
            password
        )

        self.courses = []

    def enroll_course(self, course):

        if course not in self.courses:
            self.courses.append(course)

    def drop_course(self, course):

        if course in self.courses:
            self.courses.remove(course)

    def list_courses(self):

        return [course.course_name for course in self.courses]

```

```python

# ==========================================================
# INSTRUCTOR CLASS
# ==========================================================

class Instructor(User):
    """
    Instructor inherits from User.

    Responsible for managing
    assigned courses.
    """

    def __init__(self,
                 user_id,
                 name,
                 email,
                 password):

        super().__init__(
            user_id,
            name,
            email,
            password
        )

        self.courses_taught = []

    def add_course(self, course):

        self.courses_taught.append(course)

    def remove_course(self, course):

        if course in self.courses_taught:
            self.courses_taught.remove(course)

    def list_courses(self):

        return [
            course.course_name
            for course in self.courses_taught
        ]
```

```python

# ==========================================================
# COURSE CLASS
# ==========================================================

class Course:
    """
    Represents a training course.
    """

    def __init__(self,
                 course_id,
                 course_name,
                 instructor):

        self.course_id = course_id
        self.course_name = course_name
        self.instructor = instructor

        self.learners = []

    def add_learner(self, learner):

        if learner not in self.learners:
            self.learners.append(learner)

    def remove_learner(self, learner):

        if learner in self.learners:
            self.learners.remove(learner)

    def list_learners(self):

        return [
            learner.name
            for learner in self.learners
        ]

    def display_course(self):

        return (
            f"Course ID: {self.course_id}\n"
            f"Course Name: {self.course_name}\n"
            f"Instructor: {self.instructor.name}"
        )
```

```python

# ==========================================================
# ENROLLMENT CLASS
# ==========================================================

class Enrollment:
    """
    Represents learner-course relationship.
    """

    def __init__(self,
                 enrollment_id,
                 learner,
                 course):

        self.enrollment_id = enrollment_id
        self.learner = learner
        self.course = course

    def enroll(self):
        """
        Enroll learner into course.
        """

        self.learner.enroll_course(self.course)
        self.course.add_learner(self.learner)

    def drop(self):
        """
        Remove learner from course.
        """

        self.learner.drop_course(self.course)
        self.course.remove_learner(self.learner)

    def display_enrollment(self):

        return (
            f"Enrollment ID: {self.enrollment_id}\n"
            f"Learner: {self.learner.name}\n"
            f"Course: {self.course.course_name}"
        )
```

```python

# ==========================================================
# SLTECH BACKEND
# ==========================================================

class SLTechBackend:
    """
    Central backend system.

    Stores:
    - Users
    - Courses
    - Enrollments
    """

    def __init__(self):

        self.users = []
        self.courses = []
        self.enrollments = []

    def add_user(self, user):

        self.users.append(user)

    def add_course(self, course):

        self.courses.append(course)

    def add_enrollment(self, enrollment):

        self.enrollments.append(enrollment)

    def get_enrolled_learners(self, course):

        return course.list_learners()

    def get_courses_of_learner(self, learner):

        return learner.list_courses()

    def display_all_users(self):

        print("\nALL USERS")

        for user in self.users:
            print(user.display_user())

    def display_all_courses(self):

        print("\nALL COURSES")

        for course in self.courses:
            print(course.display_course())
            print("-" * 40)


# ==========================================================
# DRIVER PROGRAM
# ==========================================================

backend = SLTechBackend()

# ==========================================================
# CREATE INSTRUCTOR
# ==========================================================

instructor1 = Instructor(
    1,
    "John Mensah",
    "john@sltech.com",
    "admin123"
)

backend.add_user(instructor1)

# ==========================================================
# CREATE COURSES
# ==========================================================

course1 = Course(
    101,
    "Python Programming",
    instructor1
)

course2 = Course(
    102,
    "Data Analytics",
    instructor1
)

instructor1.add_course(course1)
instructor1.add_course(course2)

backend.add_course(course1)
backend.add_course(course2)

# ==========================================================
# CREATE LEARNERS
# ==========================================================

learner1 = Learner(
    201,
    "Akbar Rajee",
    "akbar@gmail.com",
    "pass123"
)

learner2 = Learner(
    202,
    "Sarah Owusu",
    "sarah@gmail.com",
    "pass456"
)

backend.add_user(learner1)
backend.add_user(learner2)

# ==========================================================
# CREATE ENROLLMENTS
# ==========================================================

enrollment1 = Enrollment(
    1,
    learner1,
    course1
)

enrollment2 = Enrollment(
    2,
    learner2,
    course2
)

enrollment1.enroll()
enrollment2.enroll()

backend.add_enrollment(enrollment1)
backend.add_enrollment(enrollment2)

# ==========================================================
# OUTPUTS
# ==========================================================

backend.display_all_users()

backend.display_all_courses()

print("\nENROLLMENT DETAILS")
print(enrollment1.display_enrollment())
print()
print(enrollment2.display_enrollment())

print("\nLEARNERS IN PYTHON PROGRAMMING")
print(
    backend.get_enrolled_learners(
        course1
    )
)

print("\nCOURSES OF AKBAR RAJEE")
print(
    backend.get_courses_of_learner(
        learner1
    )
)

# ==========================================================
# SIMPLE USER INPUT SECTION
# ==========================================================

print("\nADD NEW LEARNER")

name = input("Enter learner name: ")
email = input("Enter learner email: ")
password = input("Enter learner password: ")

new_learner = Learner(
    300,
    name,
    email,
    password
)

backend.add_user(new_learner)

print("\nLearner added successfully!")

backend.display_all_users()
```
